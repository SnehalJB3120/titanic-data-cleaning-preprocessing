# Convert Categorical Columns

```python
test['Sex'] = test['Sex'].map({'male': 0, 'female': 1})
```

```python
train['Sex'] = train['Sex'].map({'male': 0, 'female': 1})
```

```python
test = pd.get_dummies(test, columns=['Embarked'], drop_first=True)
```

# Handle Outliers

```python
# Calculate Quartiles
Q1 = train['Fare'].quantile(0.25)
Q3 = train['Fare'].quantile(0.75)

# Calculate Interquartile Range
IQR = Q3 - Q1
```

```python
# calculate Limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
# Remove Outliers
train = train[(train['Fare'] >= lower_limit) & (train['Fare'] <= upper_limit)]
```

# Normalize Features

```python
test['Age'] = (
    test['Age'] - test['Age'].min()
) / (
    test['Age'].max() - test['Age'].min()
)

test['Fare'] = (
    test['Fare'] - test['Fare'].min()
) / (
    test['Fare'].max() - test['Fare'].min()
)
```

```python
train['Age'] = (
    train['Age'] - train['Age'].min()
) / (
    train['Age'].max() - train['Age'].min()
)

train['Fare'] = (
    train['Fare'] - train['Fare'].min()
) / (
    train['Fare'].max() - train['Fare'].min()
)
```
